```python
# Step 1: Remove duplicate rows
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
# Step 2: Drop rows with missing Customer ID
df = df.dropna(subset=['Customer ID'])
print("Shape after dropping missing Customer ID:", df.shape)
```

```python
# Step 3: Remove returns (negative Quantity)
df = df[df['Quantity'] > 0]
print("Shape after removing returns:", df.shape)
```

```python
# Step 4: Drop rows with missing Description
df = df.dropna(subset=['Description'])
print("Shape after dropping missing Description:", df.shape)
```

```python
# Step 5: Convert Customer ID to integer
df['Customer ID'] = df['Customer ID'].astype('int64')
df.head()
```

```python
# Check for unusual/non-product StockCodes
df['StockCode'].value_counts().head(20)
```

```python
(df['Price'] <= 0).sum()
```

```python
# Step 6: Remove non-product StockCodes
non_product_codes = ['POST', 'D', 'M', 'MANUAL', 'BANK CHARGES', 'DOT', 'CRUK', 'PADS', 'C2']
df = df[~df['StockCode'].isin(non_product_codes)]
print("Shape after removing non-product codes:", df.shape)
```

```python
# Step 7: Remove zero/negative price rows
df = df[df['Price'] > 0]
print("Shape after removing invalid prices:", df.shape)
```

```python
# Create TotalPrice column
df['TotalPrice'] = df['Quantity'] * df['Price']
df.head()
```

```python
# Convert InvoiceDate to datetime type
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Extract useful date parts
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['DayOfWeek'] = df['InvoiceDate'].dt.day_name()

df.head()
```

```python
# Save cleaned data to a new CSV
df.to_csv('cleaned_online_retail.csv', index=False)
print("Cleaned data saved successfully!")
```
